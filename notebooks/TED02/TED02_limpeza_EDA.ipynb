```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregamento da base original
df = pd.read_csv("../../data/raw/DATA (1).csv")
df.head()
```

## 1. Inspeção inicial

```python
print("Dimensão:", df.shape)
print("\nTipos de dados:")
print(df.dtypes)
print("\nValores ausentes:")
print(df.isna().sum())
print("\nDuplicados:", df.duplicated().sum())
print("\nIDs únicos:", df["STUDENT ID"].nunique())
```

## 2. Verificação de valores fora das faixas documentadas

```python
faixas = {
    "1": (1,3), "2": (1,2), "3": (1,3), "4": (1,5), "5": (1,2),
    "6": (1,2), "7": (1,2), "8": (1,5), "9": (1,4), "10": (1,4),
    "11": (1,6), "12": (1,6), "13": (1,5), "14": (1,3), "15": (1,6),
    "16": (1,5), "17": (1,5), "18": (1,3), "19": (1,3), "20": (1,2),
    "21": (1,3), "22": (1,3), "23": (1,3), "24": (1,3), "25": (1,3),
    "26": (1,3), "27": (1,3), "28": (1,3), "29": (1,5), "30": (1,5),
    "COURSE ID": (1,9), "GRADE": (0,7)
}

problemas = {}
for coluna, (minimo, maximo) in faixas.items():
    invalidos = df.loc[(df[coluna] < minimo) | (df[coluna] > maximo), coluna]
    if not invalidos.empty:
        problemas[coluna] = invalidos.tolist()

print("Valores inválidos encontrados:", problemas)
```

## 3. Padronização dos nomes das colunas

Os códigos originais foram preservados nos dados, mas os nomes das colunas foram substituídos por nomes descritivos para melhorar a leitura e a reprodutibilidade.

```python
nomes = {
    "STUDENT ID":"student_id",
    "1":"idade","2":"sexo","3":"tipo_ensino_medio","4":"tipo_bolsa","5":"trabalho_adicional",
    "6":"atividade_artistica_esportiva","7":"possui_parceiro","8":"faixa_renda","9":"transporte_universidade",
    "10":"tipo_moradia","11":"escolaridade_mae","12":"escolaridade_pai","13":"numero_irmaos",
    "14":"situacao_pais","15":"ocupacao_mae","16":"ocupacao_pai","17":"horas_estudo_semanais",
    "18":"leitura_nao_cientifica","19":"leitura_cientifica","20":"seminarios_conferencias",
    "21":"impacto_projetos_atividades","22":"frequencia_aulas","23":"preparacao_prova_companhia",
    "24":"preparacao_prova_momento","25":"anotacoes_aula","26":"atencao_aula",
    "27":"discussao_interesse_sucesso","28":"sala_invertida","29":"gpa_semestre_anterior",
    "30":"gpa_esperado_formatura","COURSE ID":"course_id","GRADE":"grade"
}
dados = df.rename(columns=nomes).copy()
dados.head()
```

## 4. Estatísticas descritivas

```python
dados.describe().T
```

## 5. Distribuição da nota final

```python
rotulos_nota = {0:"Fail",1:"DD",2:"DC",3:"CC",4:"CB",5:"BB",6:"BA",7:"AA"}
contagem = dados["grade"].value_counts().sort_index()

plt.figure(figsize=(8,4.5))
plt.bar([rotulos_nota[i] for i in contagem.index], contagem.values)
plt.title("Distribuição das notas finais")
plt.xlabel("Classe de desempenho")
plt.ylabel("Número de estudantes")
plt.show()

print(pd.DataFrame({
    "quantidade": contagem,
    "percentual": (contagem/len(dados)*100).round(1)
}))
```

## 6. Horas semanais de estudo e desempenho

```python
rotulos_estudo = {1:"Nenhuma",2:"<5 h",3:"6-10 h",4:"11-20 h",5:">20 h"}
resumo_estudo = dados.groupby("horas_estudo_semanais")["grade"].agg(["count","mean","median"])
display(resumo_estudo)

plt.figure(figsize=(8,4.5))
plt.bar([rotulos_estudo[i] for i in resumo_estudo.index], resumo_estudo["mean"])
plt.title("Média da nota final por horas semanais de estudo")
plt.xlabel("Horas semanais de estudo")
plt.ylabel("Média da classe de nota (0 a 7)")
plt.show()
```

## 7. Frequência às aulas e desempenho

```python
rotulos_frequencia = {1:"Sempre",2:"Às vezes",3:"Nunca"}
resumo_freq = dados.groupby("frequencia_aulas")["grade"].agg(["count","mean","median"])
display(resumo_freq)

plt.figure(figsize=(7,4.5))
plt.bar([rotulos_frequencia[i] for i in resumo_freq.index], resumo_freq["mean"])
plt.title("Média da nota final por frequência às aulas")
plt.xlabel("Frequência")
plt.ylabel("Média da classe de nota (0 a 7)")
plt.show()
```

## 8. Desempenho anterior e nota final

```python
rotulos_gpa = {1:"<2,00",2:"2,00-2,49",3:"2,50-2,99",4:"3,00-3,49",5:">3,49"}
resumo_gpa = dados.groupby("gpa_semestre_anterior")["grade"].agg(["count","mean","median"])
display(resumo_gpa)

plt.figure(figsize=(8,4.5))
plt.bar([rotulos_gpa[i] for i in resumo_gpa.index], resumo_gpa["mean"])
plt.title("Média da nota final por GPA do semestre anterior")
plt.xlabel("GPA do semestre anterior")
plt.ylabel("Média da classe de nota (0 a 7)")
plt.show()
```

## 9. Impacto de projetos/atividades

```python
rotulos_impacto = {1:"Positivo",2:"Negativo",3:"Neutro"}
resumo_impacto = dados.groupby("impacto_projetos_atividades")["grade"].agg(["count","mean","median"])
display(resumo_impacto)

plt.figure(figsize=(7,4.5))
plt.bar([rotulos_impacto[i] for i in resumo_impacto.index], resumo_impacto["mean"])
plt.title("Média da nota final por impacto percebido de projetos/atividades")
plt.xlabel("Impacto percebido")
plt.ylabel("Média da classe de nota (0 a 7)")
plt.show()
```

## 10. Exportação da base tratada

Nenhum registro foi removido, pois não foram encontrados valores ausentes, duplicatas nem códigos fora das faixas documentadas. A principal preparação realizada foi a padronização dos nomes das colunas. Para modelagem, `student_id` deve ser retirado por ser apenas um identificador.

```python
dados.to_csv("../../data/processed/dataset_tratado.csv", index=False)
dados.drop(columns=["student_id"]).to_csv("../../data/processed/dataset_modelagem.csv", index=False)
```
